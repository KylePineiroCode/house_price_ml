## **Encoding our Categorical Data**

I notice that we have three columns that are categorical. Those features are location, condition, and garage. I'm going to read each column to see how I would like to turn them into numerical values using different encdoing techniques. 

```python
loc_col = house_data['Location']
loc_col.unique()
```

```python
cond_col = house_data['Condition']
cond_col.unique()
```

```python
garage_col = house_data['Garage']
garage_col.unique()
```

Since the category 'Location' may have very small affect on the cost of price and living in one place is not better than the other I will use a OneHotEncoder us scikit-learn. 

```python

encoder = OneHotEncoder(sparse_output=False)
location_data = encoder.fit_transform(house_data[['Location']])
location_df = pd.DataFrame(location_data, columns=encoder.get_feature_names_out(['Location']))
house_data = pd.concat([house_data, location_df], axis=1)
house_data.head()
```

For the Condition of the house we see there are 4 different conditions. It is basically in order from 'Excellent' to 'Poor'. I am going to use ordinal encooding to make sure the value matches in a specific order. 0 = Excellet, 1 = Good, 2 = Fair, 3 = Poor. 

```python
categories_order = ['Excellent', 'Good', 'Fair', 'Poor']
encoder = OrdinalEncoder(categories=[categories_order])
house_data['Condition_Encoded'] = encoder.fit_transform(house_data[['Condition']])
house_data['Condition_Encoded'] = house_data['Condition_Encoded'].astype(int)
print(house_data[['Condition', 'Condition_Encoded']].head())
house_data.head()
```

For the garage I will do a simple mapping because it is only 'Yes' or 'No'. A house having a garage will affect the price of a house.

```python
house_data['Garage_Encoded'] = house_data['Garage'].map({'No': 0, 'Yes': 1})
house_data.head()
```
